# Calculer la moyenne

En Python, `Saisir(n)` devient un paramètre de fonction (plus facile à tester), Ecrire devient `print`, et Retourner devient `return`

```python
def moyenne_n(n):
    """Version ITERATIVE. Moyenne des entiers 1 ...n
    Traduction de l'algorithme MoyenneN(n) : Saisir(n) de vient un paramètre
    Retourne None si n = 0 (cas d'erreur) 
    """
    if n ==  0:
        print(" Erreur : n doit être supérieur à 0")
        return None
    somme = 0 
    for i in range (1, n+1):  #Pour i = 1, n faire
        somme = somme + i
    #print("Somme =", somme)
    moyenne = somme / n
    return moyenne 

n = 5
print("La moyenne des", n, "premiers nombres est:")
moyenne_n(n)
```

# Calculer la somme des _n_ premiers nombres

```python
def somme_n_rec(n):
    """Version RECURSIVE de la somme 1 + 2+ ... + n"""
    if n ==0:
        return 0
    return somme_n_rec(n-1) + n  # cas recursif

def moyenne_n_rec(n):
    """Moyenne des entiers 1..n, calculée à partir de la somme récursive"""
    if n == 0:
        print("Erreur : n doit être supérieur à 0")
        return None
    return somme_n_rec(n)/n

n = 5
print("La moyenne des", n, "premiers nombres est :")
moyenne_n_rec(n)
```

# Calculer le factoriel

```python
def fact_n(n):
    """Version ITERATIVE de n!"""
    if n == 0:
        return 1
    facto = 1
    for i in range(1, n+1):   # pour i = 1, n faire
        facto = facto * i
    return facto

n = 6
print("Le factoriel de", n, "est :", fact_n(n))

def fact_rec(n):
    """Version RECURSIVE de n!"""
    if n == 0:
        return 1     # cas de base
    return fact_rec(n-1)*n    # cas recursif 

n = 6 
print("Le factoriel de", n, "est :", fact_rec(n))
```

# Trouver l'indice _n_ de la suite de Fibonacci

```python
def fib_rec(n):
    """Version RECURSIVE (algortihme Fib_Rec)"""
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fib_rec(n-1)+fib_rec(n-2)

n = 4
print("Le fibonacci d'indice", n, "est (rec):", fib_rec(n))

def fib_iter(n):
    """Version ITERATIVE, en 0(n)"""
    if n == 0:
        return 0 
    a, b = 0, 1
    for _ in range (2, n + 1):
        a, b = b, a +b
    return b

def fib_memo(n, memo=None):
    """Version RECURSIVE MEMOIRISEE : on garde les résultats déjà calculés"""
    if memo is None:
        memo = {}
    if n in memo:
        return memo[n]
    if n == 0:
        return 0 
    if n == 1:
        return 1
    memo[n] = fib_memo(n - 1, memo) + fib_memo(n - 2, memo)
    return memo[n]
n = 10 
print("Le fibonacci d'indice", n, "est (rec memo) :", fib_memo(n))
```

# Calculer la somme de deux entiers en base 2

```python
def PlusBase2(n, m):
    z = n + m
    reste = z - ((n+m)//2)*2
    return reste

n = 3
m = 6
print("La somme de", n, "et", m," en base 2 est:", PlusBase2(n,m))
```

```python
def PlusBase2_mod(n,m):
    z = n + m
    reste_mod = z % 2
    return reste_mod

n = 3
m = 6
print("La somme de", n, "et", m," en base 2 est:", PlusBase2_mod(n,m))
```
